# Now we removing the outliers

```python
num_col
```

```python
# Code for Outlier

for i in num_col:

    # Q1 -25%
    Q1 =df[i].quantile(0.25)

    # Q3 - 75%
    Q3 =df[i].quantile(0.75)

    IQR =Q3 -Q1

    lower_bound =Q1 - 1.5 * IQR

    upper_bound =Q3 + 1.5 * IQR

    df= df[(df[i] >= lower_bound) & (df[i] <=upper_bound)]

    
```

```python
df.shape
```

```python
# We create a box plot after remove the outliers

for i in num_col:

    plt.figure(figsize =(4,3))

    sns.boxplot(x =df[i],data=df)

    plt.title(f'Boxplot After the outlier {i}')
    
```

# Model Evalution Matrix

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
```

```python
mae =mean_absolute_error(y_test,y_pred)
mse =mean_squared_error(y_test,y_pred)
rmse =root_mean_squared_error(y_test,y_pred)
r2score =r2_score(y_test,y_pred)
```

```python
print("MAE",mae)
print( "MSE",mse)
print( "RMSE",rmse)
print( "R2 -Score",r2score)
```

```python
# Training Score
lr.score(X_train,y_train)
```

```python
# Testing Score
lr.score(X_test,y_test)
```

```python
# Lasso L1
from sklearn.linear_model import  Lasso
```

```python
for i in range(200,1001,50):   # 200 - Start   ,   1001-stop    ,step =50
    l1 =Lasso(i)

    l1.fit(X_train,y_train)
    print("Value of Lambdha :",i)

    print("Training Score :",l1.score(X_train,y_train))

    print("Testing Score :",l1.score(X_test,y_test))
```
